```python
#Print numbers from 1 to N  
N=int(input("Enter any number"))
for i in range(1,N+1):
    print(i, end=" ")
```

```python
# Print numbers from N to 1 
N=int(input("Enter any number"))
for i in range(N,0,-1):
    print(i, end=" ")
```

```python
# Sum of first N natural numbers 
N = int(input("Enter any number"))
sum = 0
for i in range(N+1):
    sum = sum + i
print("Sum of N natural numbers", sum)
```

```python
# Print multiplication table of a number
N=int(input("Enter number which table you want"))
for i in range(1, 11):
    print(f"{N} x {i} = {N*i}")
```

```python
# Count digits in a number 
N = int(input("Enter the number need to count the digits"))
if N==0:
    count = 1
else:
    count = 0
    while N>0:
        N=N//10
        count=count+1
print(count)
```

```python
# Reverse a number 
N= int(input("Enter a number to reverse"))
print(N)
rev = 0
while N:
    temp = N%10
    rev = rev*10 +temp
    N= N//10

print(rev)
```

```python
# Check palindrome number  
N = int(input("Enter a number to check Palindrome or not"))
check1= N
rev=0
while N:
    temp = N%10
    rev = rev*10 +temp
    N = N//10

if check1==rev:
    print("Palindrome")
else:
    print("Not")
```
