```python
# Check palindrome number  
N = int(input("Enter a number to check Palindrome or not"))
check1= N
rev=0
while N:
    temp = N%10
    rev = rev*10 +temp
    N = N//10

if check1==rev:
    print("Palindrome")
else:
    print("Not")
```

```python
#Armstrong number check 
def count_digits(num):
    count=0
    while num>0:
        num = num//10
        count = count+1
    return count
a = int(input("Enter a number to check Armstrong"))
print("Input =",a)
temp = a
sum = 0
count = count_digits(a)
while temp>0:
    sum = (temp%10)**count + sum
    temp = temp//10
if sum == a:
    print("Armstrong", "Sum =", sum)
else:
    print("Not", "Sum =", sum)
```

```python
#GCD of two numbers 
a = int(input("Enter First number"))
print("Input a =", a)
b = int(input("Enter Second Number"))
print("Input b =",b)
GCD = 0
def get_factors(num):
    l=[]
    for i in range(1, num+1):
        if num%i == 0:
            l.append(i)
    return l
first=get_factors(a)
second = get_factors(b)
print("First =", first)
print("Second =", second)
for i in first:
    for j in range(len(second)):
        if i == second[j]:
            GCD = i
print("GCD =", GCD)
```
